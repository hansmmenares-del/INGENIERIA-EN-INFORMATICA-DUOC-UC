# Problema 1

Una red transmite datos a $100$ megabits por segundo. Crea una función utilizando *Python* que permita calcular la cantidad de datos transmitidos en una cierta cantidad de tiempo (en segundos). Luego utiliza la función y calcula cuántos datos se transmiten en:

* $45$ segundos
* $1,5$ minutos
* $1$ hora



```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import fsolve
# Referencia entregada por el ejercicio.
red_datos_1 = 100
tiempo_segundos_1 = 1
# 45 segundos totales.
tiempo_segundos_2 = 45
# 1.5 minutos totales.
tiempo_segundos_3 = 60 * 1.5
# 1 hora total.
tiempo_segundos_4 = 60 * 60
# Lista que recolecta los tiempos de transmision de datos que se quiere saber la velocidad. (X)
lista_tiempos = [tiempo_segundos_2, tiempo_segundos_3, tiempo_segundos_4]
# Lista que recoleta los megabits de informacion transmitidas en los diferentes tiempos. (Y)
lista_velocidades_transmision = []

def calcular_velocidad_transmision_datos() -> None:
    for i in lista_tiempos:
        velocidad_transmision = (red_datos_1 / tiempo_segundos_1) * i
        lista_velocidades_transmision.append(velocidad_transmision)
        print(f"Velocidad de transmisión de datos en tiempo '{i}' segundos: {velocidad_transmision} meegabits/segundo")

calcular_velocidad_transmision_datos()
plt.plot(lista_tiempos, lista_velocidades_transmision, 100)
plt.xlabel("Tiempo (segundos)")
plt.ylabel("Cantidad de datos")
plt.title("Velocidades de transmision de datos")
plt.grid(True)
plt.show()


def f(t):
    return t*100
print(f"A los 45 seg se transmiten {f(45)} megabits")


```

La latencia puede verse afectada por la velocidad de transmisión de los datos a través de los cables o conexiones inalámbricas, la distancia física entre los dispositivos, el tiempo que tarda un computador en procesar o reenviar los datos, etc.

En redes de comunicación, como las videoconferencias, juegos en línea y transmisión de datos en tiempo real, es deseable una baja latencia, de lo contrario pueden existir retrasos en la comunicación y afectar negativamente la experiencia del usuario.

Calcule la latencia real para una latencia estimada de:

a)   $200$ milisegundos.

b)   $149$ milisegundos.

c)   $74$ milisegundos.



```python
latencia_estimada_a = 200
latencia_estimada_b = 149
latencia_estimada_c = 74
lista = [latencia_estimada_a, latencia_estimada_b, latencia_estimada_c]
def latencia_real(lista: list[int]) -> None:
    for i in lista:
        latencia_real = (i * 20) / 100
        print(f"Latencia real en {i} ms: {latencia_real}")
latencia_real(lista)
```

$$U(t) = \frac{1.000}{1+9e^{-0.5t}}$$
